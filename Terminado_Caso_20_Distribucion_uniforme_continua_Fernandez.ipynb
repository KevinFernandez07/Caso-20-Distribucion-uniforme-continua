# Fundamento teórico

La distribución uniforme es una distribución de probabilidad que modela la igualdad de posibilidad entre los eventos en un rango determinado. Puede ser discreta o continua, dependiendo del conjunto de valores que puede tomar la variable aleatoria.

La distribución uniforme discreta se utiliza cuando la variable aleatoria solo puede tomar un número finito de valores, y la probabilidad de cada valor es igual. Por ejemplo, si lanzamos un dado justo, cada número del 1 al 6 tiene la misma probabilidad de 1/6 de aparecer.

Por otro lado, la distribución uniforme continua se utiliza cuando la variable aleatoria puede tomar cualquier valor en un intervalo continuo, y la probabilidad de cada valor es igual. Por ejemplo, si elegimos un número al azar entre 0 y 1, cada número en ese intervalo tiene la misma probabilidad de aparecer

La diferencia entre una distribución uniforme discreta y una distribución uniforme continua se encuentra en sus posibles valores.

La distribución uniforme discreta solo puede tomar algunos valores dentro de un intervalo además de que los valores de x's son enteros; en cambio, la distribución uniforme continua puede tomar cualquier valor dentro del intervalo en el que está definida incluyendo enteros y números reales.

En general, las distribuciones uniformes discretas solo pueden tomar valores enteros, mientras que las distribuciones uniformes continuas también pueden tomar valores decimales.

La distribución uniforme continua es una familia de distribuciones de probabilidad para variables aleatorias continuas, tales que para cada miembro de la familia, todos los intervalos de igual longitud en la distribución en su rango son igualmente probables. El dominio está definido por dos parámetros, $a$ y $b$, que son sus valores mínimo y máximo respectivamente.

La distribución o modelo uniforme puede considerarse como proveniente de un proceso de extracción aleatoria .El planteamiento radica en el hecho de que la probabilidad se distribuye uniformemente a lo largo de un intervalo . Así : dada una variable aleatoria continua, $x$ , definida en el intervalo $[a,b]$ de la recta real, se dice que $x$ tiene una distribución uniforme en el intervalo $[a,b]$.


## Función de densidad

$$
f(x) =
\begin{cases}
\frac{1}{b-a} &,\, \text{para }a\leq x \leq b ,\\
0&,\, \text{en cualquier otro caso }
\end{cases}
$$

La probabilidad únicamente depende del valor de $(x-a)$

En donde:

-   $F(x)$es la función de distribución o función de probabilidad acumulada

-   $x$ es la variable aleatoria uniforme

-   $a$ y $b$ son los valores del intervalo mínimo y máximo respectivamente.


La gráfica de esta función, conocida como *curva o función de densidad,* es un **rectángulo**, por ello la distribución uniforme continua se conoce también como *distribución rectangular* y es la más simple de las distribuciones continuas. [@lifeder].


## Función de probabilidad acumulada

Para calcular probabilidades se puede determinar a función de la distribución $F(X)$ o lo que es lo mismo la Función Acumulada de probabilidad de la distribución uniforme con la siguiente fórmula:

$$
P(x) = F(x) = \begin{cases}0; \text{ para }x \le a \\\frac{x-a}{b-a} \text{ para } a\le x \le b \\1 ; \text{ para }  x >b
\end{cases}
$$

## Cargar librerías


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform
```

## Cargar funciones

```python
def f_dens_uniforme(a, b, x):
  if (a <= x) and (x <= b):
    densidad = 1 / (b - a)
  else:
    densidad = 0
  return densidad

def f_visualizar_densidad_uniforme(x, a_min, amplitud, v_str):
  # Graficar la densidad de probabilidad
  if x is None:
    x = uniform.rvs(loc = a_min, scale = amplitud, size=10)
  # Se generan números ants y despues del inervalo para que visualizar
  # la densidad igual a cero de acuerdo con la fórmula
  numeros = np.sort(np.concatenate([[a_min-1, a_min + amplitud + 1, a_min, a_min+amplitud, a_min-0.0001, (a_min + amplitud + 0.0001)], x]))
  # print (numeros)
  pdf = uniform.pdf(x=numeros, loc = a_min, scale = amplitud )

  # print (pdf)
  plt.plot(numeros, pdf, label='Densidad de probabilidad')
  plt.title('Distribución Uniforme Continua; '+v_str)
  plt.xlabel("Valores de x's; variable aleatoria x")
  plt.ylabel('Densidad de probabilidad. Altura del rectángulo')
  plt.legend()
  plt.show()

def f_prob_uniforme(x, a, b):
  if x < a:
    probabilidad = 0
  elif x <= b:
    probabilidad = (x - a) / (b - a)
  else:
    probabilidad = 1
  return probabilidad

def f_prob_area_uniforme(a, b, a_min, b_max, v_str):
  n = 1000
  amplitud = (b_max - a_min)
  # Se generan n números aleatorios
  x = uniform.rvs(loc = a_min, scale = amplitud, size=n)
  numeros = np.sort(np.concatenate([[a_min - 1, a_min + amplitud + 1, a_min, a_min + amplitud, a_min-0.0001, (a_min + amplitud + 0.0001)], x]))

  # La densidad de cada valor aletorio
  pdf = uniform.pdf(x=numeros, loc = a_min, scale = amplitud )

  # Se visualizan
  plt.plot(numeros, pdf, label='Densidad de probabilidad')

  # Determinar la probabilidad con funciones python
  probabilidad = uniform.cdf(x = b, loc = a_min, scale = amplitud) - uniform.cdf(x = a, loc = a_min, scale = amplitud)


  # Rellenar área bajo la curva
  # Resaltar el área entre a y b
  plt.fill_between(numeros, pdf, where=(numeros >= a) & (numeros <= b), alpha=0.3, color='red', label=f'Área entre {a} y {b}, {probabilidad*100:.2f}%')

  plt.title('Distribución Uniforme Continua; '+v_str)
  plt.xlabel("Valores de x's; variable aleatoria x")
  plt.ylabel('Densidad de probabilidad. Altura del rectángulo')
  plt.legend(loc='upper right')
  plt.show()

def f_dens_uniforme_VE_desv_std(a, b, a_min, b_max, v_str, VE, desv_std):
  n = 1000
  amplitud = b_max - a_min
  # Se generan n números aleatorios
  x = uniform.rvs(loc = a_min, scale = amplitud, size=n)
  numeros = np.sort(np.concatenate([[a_min-1, a_min + amplitud + 1, a_min, a_min+amplitud, a_min-0.0001, (a_min + amplitud + 0.0001)], x]))

  # La densidad de cada valor aletorio
  pdf = uniform.pdf(x=numeros, loc = a_min, scale = amplitud )

  # Se visualizan
  plt.plot(numeros, pdf, label='Densidad de probabilidad')

  plt.axvline(x=VE, color='red', linestyle='--', label=f'Valor Esperado (VE) = {VE:.2f}')
  plt.axvline(x=VE + desv_std, color='blue', linestyle='--', label=f'Desviación Estándar = {desv_std:.2f}')
  plt.axvline(x=VE - desv_std, color='green', linestyle='--')

  plt.title('Distribución Uniforme Continua; '+v_str)
  plt.xlabel("Valores de x's; variable aleatoria x")
  plt.ylabel('Densidad de probabilidad. Altura del rectángulo')
  plt.legend(loc='upper right')
  plt.show()
```

### Calcular números aleatorios de distribución uniforme

Se calcular valores aleatorios de entre $10$ y $15$ y se visualiza la probabilidad del intervalo $10-15$

$$min = 10$$
$$amplitud = 5$$

```python
a_min = 10
amplitud = 5
b_max = a_min + amplitud
v_str = "Números aleatorios" # Variable que almacena un string
print("Números aleatorios en el rango ", a_min ," - ", b_max)
print ("min + amplitud = maximo; ", a_min, " + ", amplitud, " = ", b_max)
numeros_aleatorios = uniform.rvs(loc = a_min, scale = amplitud, size=10)
print (numeros_aleatorios)

f_visualizar_densidad_uniforme(numeros_aleatorios, a_min, amplitud, v_str)
```

## Ejercicio de llegada al trabajo

El tiempo en minutos que tarda una persona para ir de su casa al trabajo oscila de forma uniforme entre $20$ y $30$.

$$
\text{a_min = 20}
$$
$$
\text{b_max = 30}
$$
Si debe llegar al trabajo a las $8:00$ de la mañana,

Demuestre en Python ¿a qué hora debe salir de su casa para tener una probabilidad de $0.90$ o del $90%$ de no llegar tarde?

Si $x$ denota el tiempo en ir de casa al trabajo (en minutos), su distribución es uniforme en el intervalo $U(20 - 30)$.

Se busca el valor de a tal que $P(x≤a)=0.90$ o $F(x \le a) = 0.90$.

$$
F(x≤a) = \frac{a−20}{30−20}=0.90\therefore \frac{a−20}{10}=0.90∴
$$
$$
a - 20 = 0.90 * 10 \therefore a = 9 +20
$$

$$
a = 29
$$

Por tanto, debe salir de casa a las $7:31$ para tener la probabilidad del $90\%$ de llegar a las $8:00$ a su trabajo.

### La función inversa de la probabilidad uniforme

Se utiliza la función *uniform.ppf(q, loc=a, scale=(b - a))*



```python
a_min = 20 # minutos mínimos
b_max = 30 # minutos máximos
amplitud = b_max - a_min

x = uniform.ppf(q= 0.90, loc=a_min, scale=(b_max - a_min))
print ("Valor de x es: ", x)
print ("Representa los minutos antes de las 8:00 que debe salir de su casa")
```

### Comprobación y hora de salida

Se comprueba que la probabilidad acumulada en $29$ sea $0.90$ y se calcula la hora de salida restando $29$ minutos a las $8:00$.

```python
from datetime import datetime, timedelta

# Comprobando: F(x <= 29) debe ser 0.90
print ("F(x <= 29) =", uniform.cdf(x = 29, loc = a_min, scale = amplitud))
print ("Probabilidad de llegar tarde =", np.round(1 - uniform.cdf(x = 29, loc = a_min, scale = amplitud), 2))

# Hora de salida: 8:00 menos los minutos de x
hora_llegada = datetime(2026, 1, 1, 8, 0)
hora_salida = hora_llegada - timedelta(minutes = float(x))
print ("Hora de salida:", hora_salida.strftime("%H:%M"))
```
